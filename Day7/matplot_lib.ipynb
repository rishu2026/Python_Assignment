```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.random.randint(1,50,10)
y=np.random.randint(1,50,10)
print(x,y,sep="\n")
```

```python
plt.plot(x,y)
```

```python
plt.plot(x,y,color="red",linestyle="dotted")
```

```python
plt.title("Random chart")
plt.xlabel("x-axis label")
plt.ylabel("y-axis label")
plt.plot(x,y,color='red',linestyle='dotted')
plt.show()
```

```python
plt.title("Random chart")
plt.xlabel("x-axis label")
plt.ylabel("y-axis label")
plt.xticks()
plt.yticks()
plt.plot(x,y,color='red',linestyle='dotted')
plt.show()
```

```python
- -> solid(defualt)

-- -> dashed

_. -> dash-dot

: -> dotted
```

```python
plt.title("Random chart")
plt.xlabel("x-axis label")
plt.ylabel("y-axis label")
plt.xticks(x)
plt.yticks(y)
plt.plot(x,y,color='red',linestyle='-',alpha=0.3)
plt.show()
```

```python
plt.title("Random chart")
plt.xlabel("x-axis label")
plt.ylabel("y-axis label")
plt.xticks(x)
plt.yticks(y)
plt.plot(x,y,color='red',linestyle='-',alpha=0.9,marker="*")
plt.show()
```

```python
plt.title("Random chart")
plt.xlabel("x-axis label")
plt.ylabel("y-axis label")
plt.xticks(x)
plt.yticks(y)
plt.plot(x,y,color='red',linestyle='-',alpha=0.9,marker="*",markerfacecolor="b")
plt.show()
```

```python
x=np.arange(1,11)
y=np.sin(x)
z=np.cos(y)
```

```python
plt.plot(x,y,label="Sin Theta")
plt.plot(x,z,label="Cos Theta")
plt.legend()
plt.title=("Sin_Theta & Cos_Theta")
plt.xlabel("Data Points")
plt.ylabel("Associate value")
plt.xticks()
plt.yticks()
```

```python
plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.plot(x,y,label="Sin Theta",color="red")
plt.plot(x,z,label="Cos Theta")
plt.legend()
plt.title=("Sin_Theta & Cos_Theta")
plt.xlabel("Data Points")
plt.ylabel("Associate value")
plt.xticks()
plt.yticks()

plt.subplot(1,2,2)
plt.plot(x,y,label="Sin Theta",color="green")
plt.plot(x,z,label="Cos Theta",color="red")
plt.legend()
plt.title=("Sin_Theta & Cos_Theta")
plt.xlabel("Data Points2")
plt.ylabel("Associate value2")
plt.xticks()
plt.yticks()


```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

# Position 1
plt.subplot(3,3,1)
plt.plot(x, y, label="Sin Theta", color="red")
plt.plot(x, z, label="Cos Theta", color="blue")
plt.legend()

# Position 5 (center)
plt.subplot(3,3,5)
plt.plot(x, y, label="Sin Theta", color="green")
plt.plot(x, z, label="Cos Theta", color="orange")
plt.legend()

# Position 9 (bottom-right)
plt.subplot(3,3,9)
plt.plot(x, y, label="Sin Theta", color="purple")
plt.plot(x, z, label="Cos Theta", color="black")
plt.legend()

plt.tight_layout()
plt.show()
```
